# **data cleaning/removing irrelevant data and handling missing values**

```python
df_clean = df.drop(columns = ['has_photo','time','currency','source', 'id', 'price_display',])
df_clean
```

```python
df_clean.isnull().sum()/df.shape[0]*100
```

```python
df_fix = df_clean
```

```python
# Fill missing values first
df_fix["amenities"] = df_fix["amenities"].fillna("missing")

# Remove extra spaces after commas
df_fix["amenities"] = df_fix["amenities"].str.replace(", ", ",", regex=False)

# Split amenities and count each one separately
#amenity_counts = df_fix["amenities"].str.get_dummies(sep=",").sum().sort_values(ascending=False)

#amenity_counts

df_fix["amenities_clean"] = df_fix["amenities"].str.replace(", ", ",", regex=False)
df_fix["amenity_count"] = df_fix["amenities_clean"].apply(
    lambda x: 0 if x == "missing" else len(x.split(","))
)

# View the result
df_fix[["amenities", "amenity_count"]].head(10)
```

```python
# Start with a fresh copy of df_clean for df_fix to avoid previous modifications interfering
df_fix = df_clean.copy()

# 1. Create a missing indicator from the original 'amenities' column
df_fix["amenities_was_missing"] = df_fix["amenities"].isna().astype(int)

# 2. Fill original NaNs in 'amenities' with an empty string for safe processing
df_fix["amenities_processed"] = df_fix["amenities"].fillna("")

# 3. Clean spaces after commas in the processed amenities
df_fix["amenities_processed"] = df_fix["amenities_processed"].str.replace(", ", ",", regex=False)

# 4. Convert each amenity into its own binary column
# This will create columns like 'AC', 'Dishwasher', and potentially '' if there were empty strings, or 'missing' if 'missing' was a value.
amenity_encoded = df_fix["amenities_processed"].str.get_dummies(sep=",")

# 5. Remove any generated dummy columns that are not actual amenities (e.g., from empty strings or 'missing' placeholder)
amenity_encoded = amenity_encoded.drop(columns=[""], errors="ignore") # Remove if empty string generated a dummy
amenity_encoded = amenity_encoded.drop(columns=["missing"], errors="ignore") # Remove if 'missing' placeholder generated a dummy

# 6. Add prefix to make column names clear (e.g., 'AC' -> 'amenity_AC')
amenity_encoded = amenity_encoded.add_prefix("amenity_")

# 7. Calculate how many amenities each apartment has (excluding the 'missing' state itself)
df_fix["amenity_count"] = df_fix["amenities_processed"].apply(
    lambda x: 0 if x == "" or x == "missing" else len(x.split(","))
)

# 8. Add the encoded columns and the 'amenities_was_missing' indicator to df_fix
df_fix = pd.concat([df_fix, amenity_encoded, df_fix["amenities_was_missing"]], axis=1)

# 9. Drop the original 'amenities' column and the temporary 'amenities_processed' column
df_fix = df_fix.drop(columns=["amenities", "amenities_processed"]) # Drop original and temp columns
```

```python
df_fix["bedrooms"] = df_fix["bedrooms"].fillna(df["bedrooms"].median())
df_fix["bathrooms"] = df_fix["bathrooms"].fillna(df["bathrooms"].median())
df_fix["latitude"] = df_fix["latitude"].fillna(df["latitude"].median())
df_fix["longitude"] = df_fix["longitude"].fillna(df["longitude"].median())

df_fix["cityname"] = df_fix["cityname"].fillna("missing")
df_fix["state"] = df_fix["state"].fillna("missing")
```

```python
print(df_fix["cityname"].nunique())
print(df_fix["state"].nunique())
```

```python
top_cities = df_fix["cityname"].value_counts().head(50).index

# Group less common cities as Other
df_fix["cityname_grouped"] = df_fix["cityname"].where(
    df_fix["cityname"].isin(top_cities),
    "Other"
)

# Identify columns to encode, only if they still exist in the DataFrame
cols_to_encode = []
if "cityname_grouped" in df_fix.columns:
    cols_to_encode.append("cityname_grouped")
if "state" in df_fix.columns: # Check if 'state' still exists
    cols_to_encode.append("state")

# One-hot encode grouped city and all states
# Only proceed if there are columns to encode, preventing error if both are already processed
if cols_to_encode:
    df_fix = pd.get_dummies(
        df_fix,
        columns=cols_to_encode,
        drop_first=False,
        dtype=int
    )

# Drop original cityname column as it's no longer needed after creating cityname_grouped
# This assumes 'cityname' itself is not used after grouping and encoding.
# If 'cityname' was already dropped, this line will be ignored with errors='ignore'.
df_fix = df_fix.drop(columns=["cityname"], errors="ignore")
```
