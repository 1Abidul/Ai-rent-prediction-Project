```python
%%writefile app.py

import streamlit as st
import pandas as pd
import joblib

model = joblib.load("rent_project_random_forest_model.pkl")
model_columns = joblib.load("model_columns.pkl")

st.title("RentWise AI: Apartment Rent Prediction Dashboard")

st.write("""
This app predicts estimated monthly apartment rent using a trained Random Forest model.
Enter apartment details below and click Predict Rent.
""")

# Use text_input instead of number_input to avoid localtunnel widget loading errors
bedrooms = st.text_input("Bedrooms", "2")
bathrooms = st.text_input("Bathrooms", "1")
square_feet = st.text_input("Square Feet", "900")

if st.button("Predict Rent"):
    try:
        bedrooms = float(bedrooms)
        bathrooms = float(bathrooms)
        square_feet = float(square_feet)

        input_data = pd.DataFrame({
            "bedrooms": [bedrooms],
            "bathrooms": [bathrooms],
            "square_feet": [square_feet]
        })

        input_data = input_data.reindex(columns=model_columns, fill_value=0)

        prediction = model.predict(input_data)[0]

        st.subheader("Predicted Monthly Rent")
        st.success(f"${prediction:,.2f}")

        st.write("This is an estimated rent amount, not a final pricing decision.")

    except:
        st.error("Please enter valid numbers for bedrooms, bathrooms, and square feet.")

st.divider()

st.subheader("Responsible AI Note")
st.write("""
This model is only a rent-estimation support tool. Rent prices can change over time,
and the dataset may not represent every rental market equally. Users should compare
the prediction with current real-world apartment listings.
""")
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import pandas as pd
import joblib

model = joblib.load("rent_project_random_forest_model.pkl")
model_columns = joblib.load("model_columns.pkl")

# Find possible state and city columns from the trained model
state_cols = [col for col in model_columns if col.lower().startswith("state_")]
city_cols = [col for col in model_columns if col.lower().startswith("cityname_") or col.lower().startswith("city_")]

# Extract clean names for dropdowns
states = sorted([col.replace("state_", "") for col in state_cols])
cities = sorted([col.replace("cityname_", "").replace("city_", "") for col in city_cols])

# If no state/city columns exist, use default options
if len(states) == 0:
    states = ["FL", "CA", "TX", "NY"]

if len(cities) == 0:
    cities = ["Miami", "Orlando", "Tampa", "Boca Raton"]

# Input widgets
bedrooms_w = widgets.FloatText(value=2, description="Bedrooms:")
bathrooms_w = widgets.FloatText(value=1, description="Bathrooms:")
sqft_w = widgets.FloatText(value=900, description="Sq Ft:")

state_w = widgets.Dropdown(
    options=states,
    value=states[0],
    description="State:"
)

city_w = widgets.Dropdown(
    options=cities,
    value=cities[0],
    description="City:"
)

button = widgets.Button(description="Predict Rent")
output_box = widgets.Output()

def predict_rent(b):
    with output_box:
        clear_output()

        # Start with numeric inputs
        input_data = pd.DataFrame({
            "bedrooms": [bedrooms_w.value],
            "bathrooms": [bathrooms_w.value],
            "square_feet": [sqft_w.value]
        })

        # Add all model columns and fill missing ones with 0
        input_data = input_data.reindex(columns=model_columns, fill_value=0)

        # Turn on selected state column if it exists
        state_col = "state_" + state_w.value
        if state_col in input_data.columns:
            input_data[state_col] = 1

        # Turn on selected city column if it exists
        city_col_1 = "cityname_" + city_w.value
        city_col_2 = "city_" + city_w.value

        if city_col_1 in input_data.columns:
            input_data[city_col_1] = 1
        elif city_col_2 in input_data.columns:
            input_data[city_col_2] = 1

        prediction = model.predict(input_data)[0]

        print(f"Predicted Monthly Rent: ${prediction:,.2f}")
        print()
        print("Inputs used:")
        print(f"Bedrooms: {bedrooms_w.value}")
        print(f"Bathrooms: {bathrooms_w.value}")
        print(f"Square Feet: {sqft_w.value}")
        print(f"State: {state_w.value}")
        print(f"City: {city_w.value}")
        print()
        print("This is an estimated rent amount, not a final pricing decision.")

button.on_click(predict_rent)

display(bedrooms_w, bathrooms_w, sqft_w, state_w, city_w, button, output_box)
```
